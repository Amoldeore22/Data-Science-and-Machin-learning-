```python
## for loop
for i in range(5):
    print(i)
```

```python
for i in range(1,5):
    print(i)
```

```python
## range (start,stop,step)
range(1,10,2)
for i in range(1,10,2):
    print(i)
```

```python
for i in range(5,1,-1):
    print(i)
```

```python
## strings

str="Amol Deore"
for i in str:
    print(i)
```

```python
## while loop
## The while loop continue to execute as long as the condition is true

count=0
while count<5:
    print(count)
    count=count+1
```

```python
## Loop control statments
## break:exits the loop
for i in range(10):
    if i==5:
        break
    print(i)
```

```python
## continue:skip the current iteration and continue with the next.

for i in range(10):
    if i%2==0:
        continue
    print(i)
```

```python
## pass :Is a null opration; it does nothing.
```

```python
## Nested loops 
## a loop inside a loop

for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}")
```

```python
## Example- prime number between 1 to 100
for num in range(1,101):
    if num>1:
        for i in range(2,num):
            if num%i==0:
                break
        else:
            print(num)    
```
