Conditional Statments (if, elif, else)

```python
## if statment
age=18
if age>=18:
    print("You are allowed to vote in the elections")
```

```python
##else
## The else statement exucute a block of code if the condition in if statment is false
age=16
if age>=18:
    print("You are allowed to vote in the elections")
else:
    print("You are minor")
```

```python
## elif
## Allows you to check multiple condition
age=15
if age<15:
    print("You are a child")
elif age<18:
    print("you are a teenager")
else:
    print("You are an adult")
```

```python
## Nested Coditional Statment
# You can place one or more if,elif or else statement inside another if,elif or else statement to crate nested conditional statment.

num=int(input("Enter the number"))

if num>0:
    print("the numbber is positive")
    if num%2==0:
       print("The number is even") 
    else:
        print("The number is odd")

else:
    print("The number is zero or negative")
```

```python
## Determine if a year is a leap year using nested condition statement

year=int(input("Enter the year"))

if year%4==0:
    if year%100==0:
        if year%400==0:
            print(year,"is a leap year")
        else:
            print(year,"is not a leap year")
    else:
        print(year,"is a leap year")
else:
    print(year,"is not a leap year")
```

```python
## Assignment
#calculator

operation = input("Enter operation (+, -, *, /)")
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

if operation == '+':
    result = num1 + num2
elif operation == '-':
    result = num1 - num2
elif operation == '*':
    result = num1 * num2
elif operation == '/':
    if num2 == 0:
        result = "Error: Division by zero"
    else:
        result = num1 / num2
else:
    result = "Invalid operation selected."

print(f"Result: {result}")
```
